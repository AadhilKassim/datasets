```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, r2_score
```

```python
diabetes = pd.read_csv('/diabetes.csv')
diabetes.columns

```

```python
diabetes.info()
```

```python
X = diabetes.drop('Outcome', axis=1) # All columns except 'Outcome' are features
y = diabetes['Outcome'] # 'Outcome' is the target variable
feature_names = X.columns # Get feature names from the X DataFrame
print(feature_names)
```

```python
print(X.shape)
print(y.shape)
```

```python
mean = np.mean(X, axis=0)
std = np.std(X, axis=0)

X_scaled = (X - mean) / std

print("Mean of each feature:")
print(mean)

print("\nStandard Deviation of each feature:")
print(std)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
lr = LinearRegression()
lr.fit(X_train, y_train)

y_pred_lr = lr.predict(X_test)

mse_lr = mean_squared_error(y_test, y_pred_lr)
r2_lr = r2_score(y_test, y_pred_lr)
print(mse_lr)
print(r2_lr)
```

```python
ridge = Ridge()

ridge_params = {
    'alpha': [0.01, 0.1, 1, 10, 100]
}

ridge_cv = GridSearchCV(
    ridge,
    ridge_params,
    cv=5,
    scoring='neg_mean_squared_error'
)

ridge_cv.fit(X_train, y_train)

y_pred_ridge = ridge_cv.predict(X_test)

mse_ridge = mean_squared_error(y_test, y_pred_ridge)
r2_ridge = r2_score(y_test, y_pred_ridge)
print(mse_ridge)
print(r2_ridge)
```

```python
lasso = Lasso(max_iter=10000)

lasso_params = {
    'alpha': [0.001, 0.01, 0.1, 1, 10]
}

lasso_cv = GridSearchCV(
    lasso,
    lasso_params,
    cv=5,
    scoring='neg_mean_squared_error'
)

lasso_cv.fit(X_train, y_train)

y_pred_lasso = lasso_cv.predict(X_test)

mse_lasso = mean_squared_error(y_test, y_pred_lasso)
r2_lasso = r2_score(y_test, y_pred_lasso)
print(mse_lasso)
print(r2_lasso)
```

```python
results = pd.DataFrame({
    'Model': ['Linear Regression', 'Ridge Regression', 'Lasso Regression'],
    'Best Alpha': ['-',
                   ridge_cv.best_params_['alpha'],
                   lasso_cv.best_params_['alpha']],
    'MSE': [mse_lr, mse_ridge, mse_lasso],
    'R2 Score': [r2_lr, r2_ridge, r2_lasso]
})

print("\nBest Ridge Alpha:", ridge_cv.best_params_['alpha'])
print("Best Lasso Alpha:", lasso_cv.best_params_['alpha'])

print("\nPerformance Comparison")
print(results)
```
