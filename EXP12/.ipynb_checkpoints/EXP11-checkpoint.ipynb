```python
import kagglehub
path = kagglehub.dataset_download("zalando-research/fashionmnist")
```

```python
import pandas as pd
import numpy as np
import os

# Locate the CSV files in the downloaded directory
csv_files = []
for root, dirs, files in os.walk(path):
    for file in files:
        if file.endswith('.csv'):
            csv_files.append(os.path.join(root, file))

print("Found CSV files:", csv_files)

# Fallback to keras datasets if no CSV is found, else load CSVs
if len(csv_files) >= 2:
    train_path = [f for f in csv_files if 'train' in f.lower()][0]
    test_path = [f for f in csv_files if 'test' in f.lower()][0]

    train_df = pd.read_csv(train_path)
    test_df = pd.read_csv(test_path)

    print("\n--- Train Dataset Info ---")
    print(train_df.info())
    print("\n--- Train Dataset Statistical Summary ---")
    print(train_df.describe())

    y_train_raw = train_df.iloc[:, 0].values
    x_train_raw = train_df.iloc[:, 1:].values
    y_test_raw = test_df.iloc[:, 0].values
    x_test_raw = test_df.iloc[:, 1:].values
else:
    print("No CSVs found in path. Loading from Keras as fallback with equivalent structure.")
    from tensorflow.keras.datasets import fashion_mnist
    (x_train_3d, y_train_raw), (x_test_3d, y_test_raw) = fashion_mnist.load_data()
    x_train_raw = x_train_3d.reshape(-1, 784)
    x_test_raw = x_test_3d.reshape(-1, 784)
```

```python
from tensorflow.keras.utils import to_categorical

# Normalize features
X_train = x_train_raw / 255.0
X_test = x_test_raw / 255.0

# One-hot encode targets
Y_train = to_categorical(y_train_raw, 10)
Y_test = to_categorical(y_test_raw, 10)

print(f"X_train shape: {X_train.shape}, Y_train shape: {Y_train.shape}")
print(f"X_test shape: {X_test.shape}, Y_test shape: {Y_test.shape}")
```

```python
import time
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

results = {}

# Define activations to compare
activations = ['sigmoid', 'relu', 'tanh']

for act in activations:
    print(f"\nTraining network with {act.upper()} activation...")

    # Design the network
    model = Sequential([
        Dense(128, activation=act, input_shape=(784,)),
        Dense(64, activation=act),
        Dense(10, activation='softmax')
    ])

    # Compile
    model.compile(optimizer='adam',
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])

    # Train & record time
    start_time = time.time()
    history = model.fit(X_train, Y_train, epochs=8, batch_size=64, validation_split=0.1, verbose=1)
    end_time = time.time()

    training_time = end_time - start_time

    # Evaluate
    loss, accuracy = model.evaluate(X_test, Y_test, verbose=0)

    results[act] = {
        'history': history.history,
        'training_time': training_time,
        'accuracy': accuracy
    }
    print(f"{act.upper()} - Training Time: {training_time:.2f}s, Test Accuracy: {accuracy:.4f}")
```

```python
import matplotlib.pyplot as plt

# Create comparison DataFrame
comparison_df = pd.DataFrame({
    'Activation Function': [act.upper() for act in results.keys()],
    'Training Time (seconds)': [results[act]['training_time'] for act in results.keys()],
    'Classification Accuracy': [results[act]['accuracy'] for act in results.keys()]
})

display(comparison_df)

# Plotting Classification Accuracy Comparison
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.bar(comparison_df['Activation Function'], comparison_df['Classification Accuracy'], color=['blue', 'orange', 'green'])
plt.title('Classification Accuracy Comparison')
plt.xlabel('Activation Function')
plt.ylabel('Accuracy')
plt.ylim(0, 1.0)
for i, val in enumerate(comparison_df['Classification Accuracy']):
    plt.text(i, val + 0.02, f"{val:.4f}", ha='center')

# Plotting Training Time Comparison
plt.subplot(1, 2, 2)
plt.bar(comparison_df['Activation Function'], comparison_df['Training Time (seconds)'], color=['blue', 'orange', 'green'])
plt.title('Training Time Comparison')
plt.xlabel('Activation Function')
plt.ylabel('Training Time (s)')
for i, val in enumerate(comparison_df['Training Time (seconds)']):
    plt.text(i, val + 0.1, f"{val:.2f}s", ha='center')

plt.tight_layout()
plt.show()
```
